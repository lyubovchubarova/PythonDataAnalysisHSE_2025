```python
# A. Умножение

s = input()

s = s.split(" ") # разбиваем по пробелам

a = int(s[0])
b = int(s[1])

print(a * b)
```

```python
# B. Регулярные выражения. Проверка пароля

import re

s = input()

if re.match('[a-zA-Z]', s):
	print("YES")
else:
	print("NO")
```

```python
# D. Валидация артикула

import re

s = ""

pattern = r"\d\d\d"
try:
	s = input()
except:
	pass

if re.fullmatch(pattern, s):
	print("VALID")
else:
	print("INVALID")
```

```python
# E. Начало имени файла

import re

s = ""

try:
  s = input()
except:
	pass

pattern = "[a-zA-Z0-9]"

if re.match(pattern, s):
	print("CORRECT")
else:
	print("INCORRECT")
```

```python
# F. Проверка кода подтверждения

import re

s = ""

try:
  s = input()
except:
	pass

pattern = r"[A-Z]\d\s"

if re.match(pattern, s):
	print("VALID")
else:
	print("INVALID")
```

```python
# G. Проверка расширения

import re

s = ""

try:
  s = input()
except:
	pass

pattern = r"[tdx][tdx][tdx]"

if re.fullmatch(pattern, s):
	print("TEXT")
else:
	print("OTHER")
```

```python
# H. Безопасный идентификатор

import re

s = ""

try:
  s = input()
except:
	pass

pattern = r"[^012][^012][^012][^012]"

if re.fullmatch(pattern, s):
	print("SAFE")
else:
	print("UNSAFE")
```

```python
# I. Номер телефона

import re

s = ""

try:
  s = input().strip()
except:
	pass

pattern = r"\+\d{11,13}"

if re.fullmatch(pattern, s):
	print("PHONE")
else:
	print("NOT_PHONE")
```

```python
# J. Проверка времени

import re

s = ""

try:
  s = input().strip()
except:
	pass

pattern1 = r"[01][0-9]:[0-5][0-9]"
pattern2 = r"[2][0-3]:[0-5][0-9]"

if re.fullmatch(pattern1, s) or re.fullmatch(pattern2, s):
	print("VALID_TIME")
else:
	print("INVALID_TIME")
```

```python
# N. Цензура

import re

s = ""

try:
  s = input()
except:
	pass

pattern = r"\b[a-zA-Z]{4}\b"
result = re.sub(pattern, "****", s)

print(result)
```
