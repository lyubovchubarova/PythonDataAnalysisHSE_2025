# Пошаговое введение в регулярные выражения в Python

Этот конспект постепенно знакомит с основами регулярных выражений: от самых простых совпадений до более сложных конструкций.

```python
# Для начала импортируем модуль re
import re
```

## Шаг 1: Самое простое — поиск точного совпадения

Любая обычная строка уже является регулярным выражением. Она будет искать точное совпадение с самой собой.

Используем функцию `re.match()`, которая ищет совпадение **только в начале строки**.

```python
text1 = "кот и пёс"
text2 = "мой кот и пёс"

# Ищем 'кот' в начале строки text1
result1 = re.match('кот', text1)
print(f"Поиск 'кот' в '{text1}': {'Найдено' if result1 else 'Не найдено'}")

# Ищем 'кот' в начале строки text2
result2 = re.match('кот', text2)
print(f"Поиск 'кот' в '{text2}': {'Найдено' if result2 else 'Не найдено'}")
```

## Шаг 2: Специальные символы — классы символов

Часто мы не знаем точный символ, но знаем его тип (цифра, буква и т.д.). Для этого есть специальные последовательности (классы символов).

- `.` — Любой один символ (кроме новой строки).
- `\d` — Любая цифра (0-9).
- `\w` — Любая буква, цифра или знак подчеркивания `_`.
- `\s` — Любой пробельный символ (пробел, таб `\t`, новая строка `\n`).

```python
# Ищем шаблон 'к.т' (к, любой символ, т)
print('Найдено' if re.match('к.т', 'кот') else 'Не найдено')
print('Найдено' if re.match('к.т', 'кит') else 'Не найдено')
print('Найдено' if re.match('к.т', 'каат') else 'Не найдено') # Не найдено, т.к. '.' - только ОДИН символ

print('---')

# Ищем три цифры подряд
print('Найдено' if re.match('\d\d\d', '123-abc') else 'Не найдено')
```

## Шаг 3: Свои наборы символов — квадратные скобки `[]`

Квадратные скобки позволяют указать набор символов, любой из которых может стоять на этой позиции.

- `[абв]` — найдет 'а', 'б' или 'в'.
- `[а-я]` — найдет любую русскую строчную букву (диапазон).
- `[0-9]` — любая цифра (аналог `\d`).
- `[^0-9]` — любой символ, **кроме** цифры (символ `^` в начале означает отрицание).

```python
# Ищем 'кот', 'мот' или 'пот'
print('Найдено' if re.match('[кмп]от', 'кот') else 'Не найдено')
print('Найдено' if re.match('[кмп]от', 'лот') else 'Не найдено')

print('---')

# Ищем символ, который НЕ является цифрой
print('Найдено' if re.match('[^0-9]', 'A_123') else 'Не найдено')
print('Найдено' if re.match('[^0-9]', '1_ABC') else 'Не найдено')
```

## Шаг 4: Указание количества — квантификаторы

Квантификаторы ставятся после символа или группы и указывают, сколько раз они должны повторяться.

- `+` — 1 или более раз.
- `*` — 0 или более раз.
- `?` — 0 или 1 раз (необязательный символ).
- `{n}` — ровно `n` раз.
- `{n,m}` — от `n` до `m` раз.

```python
# Ищем μία или более цифр в начале строки
print('Найдено', re.match('\d+', '123-abc'))
print('Не найдено', re.match('\d+', 'abc-123'))

print('---')

# Проверяем номер телефона из 11 цифр
# Чтобы проверить всю строку, а не только начало, используем re.fullmatch()
print('Найдено', re.fullmatch('\d{11}', '89991234567'))
print('Не найдено', re.fullmatch('\d{11}', '8999123456')) # Не 11 цифр

print('---')

# Проверяем написание 'color' или 'colour'
print('Найдено', re.match('colou?r', 'color'))
print('Найдено', re.match('colou?r', 'colour'))
```

## Шаг 5: Позиция в строке — якоря и другие функции поиска

`re.match` ограничен тем, что ищет только в начале. Что если нужно найти в любом месте?

**Функции:**
- `re.search()` — ищет **первое** совпадение в любом месте строки.
- `re.findall()` — ищет **все** совпадения и возвращает их в виде списка.

**Якоря:**
- `^` — начало строки.
- `$` — конец строки.

```python
text = "Цена: 150 руб, скидка: 20 руб"

# search() найдет первое число
result_search = re.search('\d+', text)
print(f'Первое число: {result_search.group()}') # .group() возвращает найденную строку

# findall() найдет все числа
result_findall = re.findall('\d+', text)
print(f'Все числа: {result_findall}')

print('---')

# Проверим, что строка заканчивается на 'руб'
# Такой шаблон '^\d+\sруб$' проверит, что строка состоит только из числа и 'руб' и ничего больше.
print('Найдено' if re.search('руб$', 'Цена: 100 руб') else 'Не найдено')
print('Не найдено' if re.search('руб$', 'Цена: 100 рублей') else 'Не найдено')
```

## Шаг 6: Извлечение частей — группы `()`

Круглые скобки `()` позволяют "захватить" часть совпадения. Это невероятно полезно для извлечения данных.

Результат `search` или `match` имеет метод `.group(n)`, где:
- `n=0` (или без номера) — всё совпадение.
- `n=1` — первая группа (первая пара скобок).
- `n=2` — вторая группа и т.д.

```python
text = "Имя: Иван, Возраст: 25"
# (\w+) - первая группа (имя), (\d+) - вторая группа (возраст)
result = re.search('Имя: (\w+), Возраст: (\d+)', text)

if result:
    print(f"Всё совпадение: {result.group(0)}")
    print(f"Имя (группа 1): {result.group(1)}")
    print(f"Возраст (группа 2): {result.group(2)}")
    print(f"Все группы: {result.groups()}") # .groups() возвращает кортеж со всеми группами
```

## Шаг 7: Замена и разделение

- `re.sub(шаблон, замена, строка)` — заменяет все вхождения шаблона на указанную строку.
- `re.split(шаблон, строка)` — разделяет строку по разделителю, заданному шаблоном.

```python
# Удаление лишних пробелов
text_spaces = "Это    очень  длинное   предложение"
clean_text = re.sub('\s+', ' ', text_spaces)
print(f'Было: "{text_spaces}"')
print(f'Стало: "{clean_text}"')

print('---')

# Разделение по запятой и пробелам
text_csv = "яблоки, груши,   бананы,апельсины"
items = re.split(',\s*', text_csv) # разделяем по запятой и 0 или более пробелам после нее
print(items)
```

# Итоговая шпаргалка по регулярным выражениям

| Категория       | Символ(ы)       | Описание                                                              |
|-----------------|-----------------|-----------------------------------------------------------------------|
| **Символы**     | `.`             | Любой символ (кроме `\n`)                                             |
|                 | `\d` / `\D`   | Цифра / НЕ цифра                                                      |
|                 | `\w` / `\W`   | Буква, цифра или `_` / НЕ они                                          |
|                 | `\s` / `\S`   | Пробельный символ / НЕ пробельный символ                              |
| **Наборы**      | `[abc]`         | Любой из символов: `a`, `b` или `c`                                     |
|                 | `[a-z]`         | Любой символ в диапазоне от `a` до `z`                                |
|                 | `[^abc]`        | Любой символ, КРОМЕ `a`, `b` или `c`                                  |
| **Квантиф-ры**  | `*`             | 0 или более повторений                                                |
|                 | `+`             | 1 или более повторений                                                |
|                 | `?`             | 0 или 1 повторение                                                    |
|                 | `{n}`           | Ровно `n` повторений                                                  |
|                 | `{n,}`          | `n` или более повторений                                              |
|                 | `{n,m}`         | От `n` до `m` повторений                                              |
| **Якоря**       | `^`             | Начало строки                                                         |
|                 | `$`             | Конец строки                                                          |
|                 | `\b` / `\B`   | Граница слова / НЕ граница слова                                      |
| **Группы**      | `(...)`         | Захватывающая группа (для извлечения части совпадения)                |
| **Функции**     | `re.match()`    | Ищет совпадение в **начале** строки.                                  |
|                 | `re.search()`   | Ищет **первое** совпадение в любом месте строки.                      |
|                 | `re.fullmatch()`| Проверяет, **вся ли строка** соответствует шаблону.                   |
|                 | `re.findall()`  | Находит **все** совпадения и возвращает список.                       |
|                 | `re.finditer()` | Как `findall`, но возвращает итератор объектов.                       |
|                 | `re.sub()`      | Находит и **заменяет** совпадения.                                    |
|                 | `re.split()`    | **Разделяет** строку по шаблону.                                      |
