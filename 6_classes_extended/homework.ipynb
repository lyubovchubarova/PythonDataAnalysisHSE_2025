# Задание для самостоятельной работы: Игра «Крестики-нолики»

**Тема:** Классы, наследование и статические методы в Python.
**Цель:** Реализовать классическую игру «Крестики-нолики» для двух игроков в консоли, используя принципы объектно-ориентированного программирования.

**Общее описание игры:**
Два игрока по очереди ставят свои символы ('X' и 'O') на поле 3x3. Цель — собрать три своих символа в ряд по горизонтали, вертикали или диагонали. Игра заканчивается, когда один из игроков побеждает или когда на поле не остается свободных клеток (ничья).

## Техническое Задание (ТЗ)

Вам необходимо реализовать игру, используя 3 класса.

### 1. Базовый класс `Board` (Доска)
Этот класс отвечает **только за хранение данных** о состоянии игрового поля. Он ничего не знает о том, как его отображать.
- **Атрибуты:**
  - `grid`: Двумерный список (список списков) размером 3x3, который представляет игровое поле. В начале он должен быть заполнен пустыми символами, например, пробелами `' '`.
- **Методы:**
  - `__init__(self)`: Конструктор, создающий и инициализирующий пустое поле `self.grid`.
  - `is_full(self)`: Проверяет, заполнено ли поле. Возвращает `True`, если на поле не осталось пустых клеток, иначе `False`.
  - `place_symbol(self, row, col, symbol)`: Помещает `symbol` ('X' или 'O') в указанную ячейку `(row, col)`. Перед этим проверяет, что ячейка свободна. Если ячейка уже занята, метод должен возвращать `False`, если символ успешно поставлен — `True`.

--- 
### 2. Класс `ConsoleBoard` (Консольная доска, наследуется от `Board`)
Этот класс **наследует** всю логику хранения данных от `Board` и добавляет к ней функциональность **отображения** в консоли.
- **Наследование:** Должен быть наследником класса `Board`.
- **Методы:**
  - `__init__(self)`: Конструктор, который просто вызывает конструктор родительского класса `super().__init__()`.
  - `display(self)`: Отрисовывает игровое поле в консоли в наглядном виде. Например:
    ```
     X | O |  
    ---|---|---
       | X |  
    ---|---|---
     O |   | X 
    ```

--- 
### 3. Класс `Game` (Игровой движок)
Основной класс, который управляет логикой игры, очередностью ходов и проверкой на победу.
- **Атрибуты:**
  - `board`: Экземпляр класса `ConsoleBoard`.
  - `current_player`: Строка, хранящая символ текущего игрока ('X' или 'O').
- **Методы:**
  - `__init__(self)`: Конструктор. Создает `self.board` и устанавливает `self.current_player` в 'X'.
  - `switch_player(self)`: Меняет текущего игрока с 'X' на 'O' и наоборот.
  - **`check_win(grid)` — статический метод!**
    - **Назначение:** Проверяет, есть ли на поле `grid` выигрышная комбинация. Этот метод не зависит от конкретной игры (`self`), а работает с любым переданным ему полем 3x3. Это идеальный кандидат для статического метода.
    - **Логика:** Проверяет все 8 выигрышных комбинаций (3 горизонтали, 3 вертикали, 2 диагонали). Если находит ряд из трех 'X' или трех 'O', возвращает этот символ ('X' или 'O'). Если победителя нет, возвращает `None`.
    - **Реализация:** Должен быть объявлен с помощью декоратора `@staticmethod`.
  - `play_turn(self)`: Организует один ход игрока.
    - Спрашивает у пользователя номер строки и столбца (от 1 до 3).
    - Проверяет корректность ввода.
    - Пытается поставить символ на доску с помощью `self.board.place_symbol()`.
    - Если ход невозможен (ячейка занята), сообщает об этом и просит сделать ход заново.
  - `run(self)`: Главный игровой цикл.
    - В бесконечном цикле `while True`:
      1. Отображает доску (`self.board.display()`).
      2. Организует ход текущего игрока (`play_turn`).
      3. Проверяет, победил ли текущий игрок, с помощью **статического метода** `Game.check_win(self.board.grid)`. Если да, выводит сообщение о победе и прерывает цикл (`break`).
      4. Проверяет, заполнена ли доска (`self.board.is_full()`). Если да, выводит сообщение о ничьей и прерывает цикл.
      5. Если игра не окончена, меняет игрока (`switch_player`).
