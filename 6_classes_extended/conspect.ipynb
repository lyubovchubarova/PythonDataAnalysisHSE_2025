В этом занятии мы разберем продвинутые, но очень полезные инструменты работы с классами в Python, а также освежим в памяти три фундаментальных принципа Объектно-Ориентированного Программирования (ООП).

## Часть 1: Расширенные возможности классов

### 1. Атрибуты класса vs. Атрибуты экземпляра

В Python существует два типа атрибутов, связанных с классами:

1.  **Атрибуты экземпляра (Instance Attributes)**: Эти атрибуты уникальны для каждого объекта (экземпляра) класса. Обычно они определяются внутри метода `__init__` с использованием `self`.
2.  **Атрибуты класса (Class Attributes)**: Эти атрибуты являются общими для всех экземпляров класса. Они определяются непосредственно внутри тела класса, вне каких-либо методов.

Давайте рассмотрим на примере. Создадим класс `Car`.

```python
class Car:
    # Атрибут класса: общий для всех машин счетчик
    total_cars_produced = 0
    
    # Атрибут класса: все машины ездят на колесах
    wheels = 4

    def __init__(self, brand, model):
        # Атрибуты экземпляра: у каждой машины своя марка и модель
        self.brand = brand
        self.model = model
        
        # При создании новой машины, увеличиваем общий счетчик, обращаясь к атрибуту класса
        Car.total_cars_produced += 1
        print(f"Создана новая машина: {self.brand} {self.model}. Всего произведено: {Car.total_cars_produced}")

    def display_info(self):
        print(f"Марка: {self.brand}, Модель: {self.model}, Колес: {self.wheels}")
```

Теперь создадим несколько экземпляров класса `Car`.

```python
car1 = Car("Toyota", "Camry")
car2 = Car("Ford", "Mustang")
car3 = Car("Tesla", "Model S")
```

Атрибуты `brand` и `model` уникальны для каждого автомобиля:

```python
print(f"Машина 1: {car1.brand} {car1.model}")
print(f"Машина 2: {car2.brand} {car2.model}")
```

Атрибуты `total_cars_produced` и `wheels` являются общими. К ним можно обратиться как через класс, так и через любой из его экземпляров.

```python
# Доступ через класс
print(f"Всего произведено машин (доступ через класс): {Car.total_cars_produced}")

# Доступ через экземпляры (Python сначала ищет атрибут у экземпляра, а если не находит - у класса)
print(f"Колес у car1: {car1.wheels}")
print(f"Колес у car2: {car2.wheels}")
```

**Важный момент:** Что произойдет, если мы попытаемся изменить атрибут класса через экземпляр?

```python
print(f"Изначальное кол-во колес у car1: {car1.wheels}")
print(f"Изначальное кол-во колес у car2: {car2.wheels}")
print(f"Изначальное кол-во колес в классе Car: {Car.wheels}")

# Попытаемся сделать из car2 трицикл
car2.wheels = 3
print("\n--- После изменения car2.wheels = 3 ---")

print(f"Колес у car1: {car1.wheels}") # Не изменилось
print(f"Колес у car2: {car2.wheels}") # Изменилось
print(f"Колес в классе Car: {Car.wheels}") # Не изменилось

# Что на самом деле произошло?
# У экземпляра car2 был создан СОБСТВЕННЫЙ атрибут 'wheels', который "затенил" атрибут класса.
print("\nАтрибуты экземпляра car1:", car1.__dict__)
print("Атрибуты экземпляра car2:", car2.__dict__)
```

Чтобы изменить атрибут для всех экземпляров, его нужно менять через сам класс.

```python
# Теперь все машины, у которых нет своего атрибута wheels, будут иметь новое значение
Car.wheels = 6 # Делаем все машины грузовиками

print(f"Колес у car1 (после изменения Car.wheels): {car1.wheels}")
print(f"Колес у car2 (у него свой атрибут): {car2.wheels}")

car4 = Car("Kamaz", "54901")
car4.display_info()
```

### 2. `@staticmethod` (Статический метод)

**Статический метод** — это функция, которая находится внутри класса, но не привязана ни к экземпляру (`self`), ни к самому классу (`cls`). По сути, это обычная функция, помещенная в пространство имен класса для логической группировки.

**Когда использовать?**
- Когда функция логически связана с классом, но для своей работы ей не нужны данные ни о классе, ни о его конкретном экземпляре.
- Как "вспомогательная" или утилитная функция.

Декоратор: `@staticmethod`.

```python
class MathHelper:

    @staticmethod
    def is_integer(num):
        """Проверяет, является ли число целым."""
        # Обратите внимание: здесь нет ни self, ни cls
        if isinstance(num, int):
            return True
        if isinstance(num, float) and num.is_integer():
            return True
        return False

    @staticmethod
    def circle_area(radius):
        """Вычисляет площадь круга."""
        return 3.14159 * radius ** 2
```

Статический метод можно вызывать как от имени класса, так и от экземпляра. Результат будет одинаковым.

```python
# Вызов через класс
print(f"Является ли 5 целым числом? {MathHelper.is_integer(5)}")
print(f"Является ли 5.0 целым числом? {MathHelper.is_integer(5.0)}")
print(f"Является ли 5.1 целым числом? {MathHelper.is_integer(5.1)}")
print(f"Площадь круга с радиусом 10: {MathHelper.circle_area(10)}")

# Вызов через экземпляр (менее распространен, но возможен)
# Для этого не нужно даже определять __init__
helper_instance = MathHelper()
print(f"\nВызов через экземпляр: {helper_instance.is_integer(7.0)}")
```

### 3. `@classmethod` (Метод класса)

**Метод класса** — это метод, который в качестве первого аргумента принимает сам класс (`cls`), а не экземпляр (`self`). Он может работать с атрибутами класса, но не с атрибутами экземпляра.

**Когда использовать?**
- Основное применение — создание **альтернативных конструкторов**. То есть, способов создания экземпляра класса из данных в другом формате.
- Когда методу нужен доступ к классу (например, для вызова других методов класса или доступа к атрибутам класса).

Декоратор: `@classmethod`.

```python
import datetime

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Имя: {self.name}, Возраст: {self.age}")
    
    @classmethod
    def from_birth_year(cls, name, birth_year):
        """Альтернативный конструктор, создающий экземпляр по году рождения."""
        current_year = datetime.date.today().year
        age = current_year - birth_year
        # Вместо Person(name, age) мы используем cls(name, age).
        # Это позволяет методу правильно работать в дочерних классах!
        return cls(name, age)
```

Теперь мы можем создавать объекты `Person` двумя способами.

```python
# Стандартный способ
person1 = Person("Иван", 30)
person1.display()

# Альтернативный способ через classmethod
person2 = Person.from_birth_year("Анна", 1995)
person2.display()
```

Главное преимущество `classmethod` проявляется при наследовании. Метод класса, унаследованный дочерним классом, будет создавать экземпляры дочернего класса, а не родительского.

```python
class Student(Person):
    # Переопределим display для наглядности
    def display(self):
        print(f"Студент: {self.name}, Возраст: {self.age}")

# Используем унаследованный @classmethod для создания экземпляра Student
student1 = Student.from_birth_year("Пётр", 2003)

student1.display()
print(f"Тип объекта: {type(student1)}")

# Если бы в from_birth_year мы написали 'return Person(name, age)',
# то создался бы объект типа Person, а не Student, что было бы неверно.
```

### 4. `@dataclass` (Классы данных)

Часто классы используются просто для хранения данных. В таких случаях приходится писать много шаблонного кода: `__init__`, `__repr__` (для красивого вывода), `__eq__` (для сравнения) и т.д.

Декоратор `@dataclass` (из модуля `dataclasses`) автоматизирует это. Вы просто объявляете переменные с аннотациями типов, а Python сам генерирует все нужные методы.

**Пример: "старый" способ**

```python
class PointOld:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"PointOld(x={self.x}, y={self.y})"

    def __eq__(self, other):
        if not isinstance(other, PointOld):
            return NotImplemented
        return self.x == other.x and self.y == other.y

p_old1 = PointOld(1, 2)
p_old2 = PointOld(1, 2)
print(p_old1)  # Без __repr__ был бы некрасивый вывод
print(p_old1 == p_old2) # Без __eq__ было бы False
```

**Пример: с `@dataclass`**

```python
from dataclasses import dataclass, field

@dataclass(order=True) # order=True добавит методы для сравнения (<, >, <=, >=)
class Point:
    x: int
    y: int
    z: int = 0 # Можно задать значение по умолчанию

p1 = Point(1, 2)
p2 = Point(1, 2)
p3 = Point(10, -5)

print(p1) # __repr__ сгенерирован автоматически
print(p1 == p2) # __eq__ сгенерирован автоматически
print(p1 == p3)
print(p3 > p1) # работает, так как мы указали order=True
```

**Итог по `@dataclass`**: Это современный и очень удобный способ для создания классов, основная цель которых — хранение данных. Он значительно сокращает количество шаблонного кода.

## 1. Инкапсуляция

**Инкапсуляция** — это механизм, который объединяет данные и методы, работающие с этими данными, в единый компонент (класс), и при этом **скрывает детали реализации** от внешнего мира.

+ **Цель**: Управление доступом к данным объекта. Мы не хотим, чтобы кто-то извне мог случайно или намеренно "сломать" внутреннее состояние объекта.
+ В Python уровни доступа — это скорее **соглашения**, а не строгие правила.

Уровни доступа:
- **Public** (Публичный): Имена без подчеркиваний (`my_variable`). Доступны отовсюду.
- **Protected** (Защищенный): Имена с одним нижним подчеркиванием в начале (`_my_variable`). Соглашение о том, что это внутренняя часть класса и ее не следует трогать извне. Однако технически доступ к ней есть.
- **Private** (Приватный): Имена с двумя нижними подчеркиваниями в начале (`__my_variable`). Python *переименовывает* такие атрибуты, чтобы избежать случайных конфликтов в дочерних классах. Это называется **name mangling** (искажение имени).

#### Protected атрибут (`_`)

```python
class SomeClassProtected:
    
    def __init__(self, n):
        # Этот атрибут считается защищенным
        self._protected_n = n
        
    def _protected_method(self):
        print("Это защищенный метод объекта. Не трогайте его без необходимости!")

obj = SomeClassProtected(2)

# Технически мы МОЖЕМ это сделать, но так делать НЕ РЕКОМЕНДУЕТСЯ.
# Это нарушает "контракт" класса.
obj._protected_method() 
obj._protected_n = 1 
print(obj._protected_n)
```

#### Private атрибут (`__`)

```python
class SomeClassPrivate:
    
    def __init__(self, n):
        # Этот атрибут является приватным
        self.__private_n = n
        
    def __private_method(self):
        print("Это приватный метод объекта")
    
    def use_private_stuff(self):
        print(f"Внутри класса мы можем легко использовать приватные члены: {self.__private_n}")
        self.__private_method()

obj = SomeClassPrivate(2)
obj.use_private_stuff()

try:
    # Вызов напрямую приведет к ошибке AttributeError
    obj.__private_method()
except AttributeError as e:
    print(f"\nОшибка при прямом доступе: {e}")
```

Но, зная механизм искажения имен (`_ИмяКласса__имяАтрибута`), добраться до приватного атрибута все еще можно:

```python
print("\nНо если очень хочется, то можно:")
obj._SomeClassPrivate__private_method() # вызываем внутренний метод объекта 
obj._SomeClassPrivate__private_n = 1 # изменяем внутренний атрибут объекта 
print(obj._SomeClassPrivate__private_n)
```

#### Pythonic way: Свойства (`@property`)
Более правильный способ инкапсуляции в Python — это использование свойств. Они позволяют обращаться к методу как к атрибуту, предоставляя контролируемый доступ.

- `@property`: метод-геттер (получение значения).
- `@<имя>.setter`: метод-сеттер (установка значения).

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance # Защищенный атрибут для хранения баланса
        
    @property
    def balance(self):
        # Этот метод будет вызываться, когда мы обращаемся к 'account.balance'
        print("Getter: Получение баланса")
        return f"{self._balance} USD"
    
    @balance.setter
    def balance(self, value):
        # Этот метод будет вызываться, когда мы делаем 'account.balance = ...'
        print("Setter: Попытка изменить баланс")
        if value < 0:
            print("Ошибка: баланс не может быть отрицательным!")
        else:
            self._balance = value

account = BankAccount("Иван Петров", 1000)

# Обращаемся к свойству 'balance' (вызывается геттер)
print(f"Владелец: {account.owner}, Текущий баланс: {account.balance}")

# Пытаемся изменить баланс (вызывается сеттер)
account.balance = 1500
print(f"Новый баланс: {account.balance}")

# Пытаемся установить некорректное значение (сеттер сработает и не даст этого сделать)
account.balance = -500
print(f"Баланс после некорректной операции: {account.balance}")
```

## 2. Наследование

**Наследование** — это механизм, позволяющий создавать новый класс (потомок, дочерний класс) на основе уже существующего (предок, родительский класс).

+ **Цель**: Повторное использование кода. Дочерний класс автоматически получает все атрибуты и методы родительского класса.
+ Дочерний класс может **расширять** функциональность (добавлять новые методы и атрибуты) и **переопределять** ее (изменять поведение существующих методов).

#### Одиночное наследование

```python
class Animal:
   
    fav_food = 'еда' # атрибут класса 
    
    def __init__(self, name, legs, scariness):
        self.name = name 
        self.legs = legs
        self.scariness = scariness
    
    def introduce(self): 
        print("Привет! Меня зовут %s!" % self.name)
    
    def sound(self):
        print("Издаю какой-то звук!")

class Mammal(Animal): # имя родительского класса пишется в скобках 
    
    def __init__(self, name, scariness): 
        # обращаемся к классу-родителю с помощью super() и вызываем его метод __init__
        # Это позволяет инициализировать атрибуты, определенные в родительском классе.
        super().__init__(name=name, legs=4, scariness=scariness) # пусть у всех млекопитающих должно быть 4 ноги
    
    def feed_with_milk(self):
        print(f"{self.name} кормит детенышей молоком.")
```

Класс-потомок `Mammal` наследует все методы и атрибуты родительского класса `Animal`.

```python
mammal = Mammal(name='Киса', scariness=1)

# Унаследованные атрибуты и методы
print(f"Любимая еда: {mammal.fav_food}")
print(f"Количество ног: {mammal.legs}")
mammal.introduce()
mammal.sound()

# Новый метод, который есть только у Mammal
mammal.feed_with_milk()
```

**Вопрос**: что будет, если создать потомка вот так, без `__init__`?

```python
class Reptile(Animal):
    # У этого класса нет своего __init__
    pass

# При создании экземпляра Reptile будет автоматически вызван __init__ родительского класса Animal
# Поэтому мы должны передать все аргументы, которые ожидает Animal.__init__
snake = Reptile(name='Каа', legs=0, scariness=8)
snake.introduce()
```

**Ответ**: Если в дочернем классе не определен `__init__`, то при создании объекта будет автоматически вызван `__init__` его родителя. Если и у родителя нет, то поиск продолжится вверх по иерархии наследования.

#### Множественное наследование

![](https://media.proglib.io/wp-uploads/2018/12/7660079.png)

Класс может наследовать от нескольких родительских классов. В этом случае он получает методы и атрибуты от всех них. Родители перечисляются в скобках через запятую.

```python
class Donkey:
    def __init__(self, name):
        self.name = name
    
    def make_sound(self):
        print("Иа-Иа!")
    
class Horse:
    def __init__(self, name):
        self.name = name
        
    def run(self):
        print("Скачет галопом")

class Mule(Donkey, Horse):
    # Потомок наследует от обоих классов
    pass

mule = Mule("Упрямец")
mule.make_sound() # от Donkey
mule.run()        # от Horse
```

## 3. Полиморфизм

**Полиморфизм** — это способность объектов с одинаковым интерфейсом (набором методов) иметь разную реализацию этих методов.

+ **Цель**: Создавать унифицированный код, который может работать с объектами разных классов, не зная их конкретного типа, но будучи уверенным, что у них есть нужный метод.
+ Простейший пример из Python — оператор `+`. Он работает и для чисел (сложение), и для строк (конкатенация), и для списков (слияние). Это полиморфизм в действии.

В ООП полиморфизм чаще всего достигается путем **переопределения методов** в дочерних классах.

```python
print(1 + 1)              # Сложение чисел
print("1" + "1")            # Конкатенация строк
print([1, 2] + [3, 4])  # Слияние списков

# За кулисами вызываются разные методы: int.__add__, str.__add__, list.__add__
```

Давайте вернемся к нашим животным. Родительский `Animal` имеет метод `sound`. Каждый потомок может реализовать этот метод по-своему.

```python
class Cat(Mammal):
    # переопределяем метод sound
    def sound(self): 
        print("Мяу!")
    
class Dog(Mammal):
    # переопределяем метод sound
    def sound(self): 
        print("Гав!")

class Cow(Mammal):
    # переопределяем метод sound
    def sound(self):
        print("Мууу!")

# Создаем экземпляры. Для Mammal нужен name и scariness.
cat = Cat(name='Барсик', scariness=2)
dog = Dog(name='Шарик', scariness=3)
cow = Cow(name='Мурка', scariness=1)
```

Теперь у нас есть несколько объектов разных типов, но все они имеют одинаковый интерфейс — метод `sound()`. Мы можем работать с ними единообразно, не проверяя тип каждого объекта.

```python
animals_on_farm = [cat, dog, cow]

for animal in animals_on_farm:
    print(f"{animal.name} говорит:", end=" ")
    animal.sound() # Вызывается РАЗНАЯ реализация метода sound() в зависимости от реального типа объекта
```

#### Полиморфизм и Множественное наследование: проблема Котопса

**Вопрос**: что если создать класс, который наследуется и от `Cat`, и от `Dog`? Какой звук он будет издавать?

```python
# Порядок наследования важен: Cat, Dog
class CatDog(Cat, Dog):
    pass

catdog = CatDog('Котопёс', scariness=5)
catdog.sound() # Почему он мяукает?
```

**Ответ**: При вызове метода Python ищет его реализацию в соответствии с **MRO (Method Resolution Order)** — Порядком разрешения методов. Это последовательность классов, которые Python просматривает в поиске метода.

При наследовании `class CatDog(Cat, Dog)` MRO будет таким, что сначала будет просмотрен класс `Cat` и его предки, а только потом `Dog` и его предки. Поскольку метод `sound` находится в `Cat`, он используется и поиск прекращается.

```python
# Давайте посмотрим на MRO для нашего класса
print(CatDog.mro())
```

Как заставить котопса гавкать?

1.  **Изменить порядок наследования**: `class DogCat(Dog, Cat): pass`
2.  **Явно переопределить метод** в самом классе `CatDog`.

```python
# Способ 1: меняем порядок
class DogCat(Dog, Cat):
    pass

dogcat = DogCat('Пёсокот', 4)
print("Пёсокот говорит:", end=" ")
dogcat.sound()
print(DogCat.mro())

# Способ 2: переопределяем метод
class CatDog(Cat, Dog):
    def sound(self):
        print("Мяу-Гав!")

catdog_new = CatDog('Новый Котопёс', 5)
print("Новый Котопёс говорит:", end=" ")
catdog_new.sound()
```
