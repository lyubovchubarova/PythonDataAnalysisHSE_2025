# Большой конспект: Классы и ООП в Python (v2.0)

## 1. Введение: Зачем нужны классы?

### 1.1. Теоретическая аналогия: Чертеж и Дом

Представьте, что вы строите дома. У вас есть **чертеж** дома. Этот чертеж — не дом. Вы не можете в нем жить. Но он описывает все характеристики будущего дома: сколько в нем этажей, какая площадь, сколько окон, из какого материала стены.

Используя один и тот же чертеж, вы можете построить множество **реальных домов**. Каждый дом будет отдельным, самостоятельным объектом, но все они будут созданы по одному и тому же плану.

В программировании эта аналогия работает идеально:

*   **Класс (Class)** — это **чертеж**. Это шаблон, описывающий, какими **данными (атрибутами)** и каким **поведением (методами)** будут обладать будущие объекты. 
*   **Объект (Object)** или **Экземпляр класса (Instance)** — это **реальный дом**, построенный по чертежу. Это конкретная сущность в памяти компьютера, созданная на основе класса. У нее есть свои собственные, уникальные значения атрибутов.

**Основная идея ООП** — объединить данные и функции, которые с этими данными работают, в единую сущность — объект. Это делает код более организованным, понятным и позволяет легко его переиспользовать.

### 1.2. Объект vs. Экземпляр: Разбираемся в терминах

На практике эти термины часто используются как синонимы, но между ними есть тонкое концептуальное различие.

**Объект (Object)** — это фундаментальное понятие в Python. **Практически всё в Python является объектом**: числа, строки, списки, функции и даже сами классы. 

**Экземпляр (Instance)** — это **конкретный объект, созданный по вашему чертежу-классу**. Когда вы пишете `my_dog = Dog()`, вы создаете *экземпляр* класса `Dog`. 

> **Проще говоря:** любой экземпляр — это объект, но не каждый объект (например, число `5`) — это экземпляр *вашего* класса `Dog`.

```python
class Dog:
    pass

# my_dog - это и ОБЪЕКТ, и ЭКЗЕМПЛЯР класса Dog
my_dog = Dog()

# 5 - это тоже ОБЪЕКТ, и он является ЭКЗЕМПЛЯРОМ встроенного класса int
number = 5

# `isinstance()` - идеальная функция для проверки
print(f"my_dog - экземпляр Dog? {isinstance(my_dog, Dog)}")
print(f"number - экземпляр Dog? {isinstance(number, Dog)}")
print(f"number - экземпляр int? {isinstance(number, int)}")

# Важно: в Python все наследуется от базового класса `object`
print(f"my_dog - это объект? {isinstance(my_dog, object)}")
print(f"number - это объект? {isinstance(number, object)}")
```

## 2. Базовый синтаксис: от чертежа к объекту

Начнем с самых азов и будем постепенно усложнять наш класс.

### 2.1. Минимальный класс и создание экземпляра

Синтаксис для создания класса очень прост: ключевое слово `class`, затем имя класса (по соглашению в стиле `CamelCase`) и двоеточие. Если класс пока ничего не делает, используется заглушка `pass`.

```python
class Cat:
    pass

# Создание экземпляра (объекта) класса Cat. 
# Этот процесс называется "инстанцирование".
cat1 = Cat()
cat2 = Cat()

# cat1 и cat2 - это два независимых объекта, созданных по одному "чертежу"
print(cat1)
print(cat2)

# Они не равны друг другу и хранятся в разных местах памяти
print(f"cat1 и cat2 - это один и тот же объект? {cat1 is cat2}")
```

### 2.2. Конструктор `__init__` и Атрибуты экземпляра

Наши коты пока бесполезны — у них нет ни имени, ни возраста. Чтобы при создании объекта задать ему уникальные характеристики (данные), используется **конструктор** — специальный метод `__init__`.

*   `__init__` вызывается **автоматически** в момент создания экземпляра (`Cat(...)`).
*   `self` — это обязательный первый параметр. Это ссылка **на сам создаваемый объект**. Именно через `self` мы "прикрепляем" к объекту его данные.
*   `self.имя_атрибута` — это **атрибут экземпляра**. Он принадлежит конкретному объекту и будет уникальным для каждого кота.

```python
class Cat:
    # Конструктор класса
    def __init__(self, name, age, color):
        print(f"Создается новый кот! Его зовут {name}.")
        # Создаем атрибуты экземпляра
        self.name = name
        self.age = age
        self.color = color

# Теперь при создании мы обязаны передать аргументы в __init__
cat1 = Cat("Барсик", 3, "Рыжий")
cat2 = Cat("Мурка", 5, "Черепаховый")

# Доступ к атрибутам осуществляется через точку
print(f"Имя первого кота: {cat1.name}, его цвет: {cat1.color}")
print(f"Имя второго кота: {cat2.name}, его возраст: {cat2.age}")

# Атрибуты можно изменять
cat1.age = 4
print(f"Барсику исполнилось {cat1.age} года!")
```

### 2.3. Методы: добавляем поведение

Объекты должны не только хранить данные, но и что-то *делать*. Функции, определенные внутри класса, называются **методами**. Они определяют поведение объекта.

Методы экземпляра, как и `__init__`, **всегда принимают `self` первым аргументом**. Это позволяет им получать доступ к атрибутам (`self.name`, `self.age`) конкретного объекта, который их вызвал.

```python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.is_hungry = True

    # Метод экземпляра №1
    def meow(self):
        print(f"{self.name} говорит: Мяу-мяу!")

    # Метод экземпляра №2, который меняет состояние объекта
    def feed(self):
        if self.is_hungry:
            print(f"{self.name} с аппетитом ест... Муррр. Теперь он сыт.")
            self.is_hungry = False
        else:
            print(f"{self.name} уже сыт и отказывается от еды.")

barsik = Cat("Барсик", 3)

# Вызываем методы для нашего экземпляра
barsik.meow()
print(f"Барсик голоден? {barsik.is_hungry}")

barsik.feed()
print(f"А теперь Барсик голоден? {barsik.is_hungry}")

barsik.feed() # Пробуем покормить еще раз
```

## 3. Глубже в атрибуты: Класс vs. Экземпляр

Мы уже знаем про **атрибуты экземпляра** (`self.name`), которые уникальны для каждого объекта. Но есть и **атрибуты класса**, которые являются общими для всех экземпляров.

*   **Атрибуты класса** определяются прямо в теле класса, вне методов. Они хранятся в единственном экземпляре в самом классе.
*   **Атрибуты экземпляра** определяются в `__init__` через `self`. Они хранятся в каждом отдельном объекте.

### Пример и "Эффект Тени"
Доступ к атрибуту класса можно получить как через сам класс (`Cat.species`), так и через экземпляр (`barsik.species`). Python сначала ищет атрибут в экземпляре. Если не находит — ищет в классе. 

Что произойдет, если мы попытаемся изменить атрибут класса через экземпляр?

```python
class Cat:
    # Атрибут класса: общий для всех котов
    species = 'Felis catus'
    paws_count = 4

    def __init__(self, name):
        # Атрибут экземпляра: уникальный для каждого
        self.name = name

barsik = Cat('Барсик')
murka = Cat('Мурка')

# Все смотрят на один и тот же атрибут класса
print(f"Вид Барсика: {barsik.species}")
print(f"Вид Мурки: {murka.species}")
print(f"Вид класса Cat: {Cat.species}")

print("\n--- Изменяем атрибут у Барсика ---")
# ВАЖНО: Эта операция не меняет атрибут класса.
# Она создает НОВЫЙ атрибут экземпляра `species` только для объекта `barsik`.
# Этот новый атрибут "затеняет" (shadows) атрибут класса.
barsik.species = 'Super Cat'

print(f"Вид Барсика: {barsik.species}") # barsik теперь использует свой собственный атрибут
print(f"Вид Мурки: {murka.species}") # murka по-прежнему использует атрибут класса
print(f"Вид класса Cat: {Cat.species}") # атрибут класса не изменился!

print("\n--- Изменяем атрибут через Класс ---")
# Чтобы изменить атрибут для всех, нужно обращаться через класс:
Cat.species = 'Felis Silvestris Catus'
print(f"Новый вид для Мурки: {murka.species}")
print(f"А у Барсика все еще свой: {barsik.species}")
```

## 4. Столпы Объектно-Ориентированного Программирования (ООП)

Теперь, когда у нас есть прочная база, мы можем рассмотреть три основных принципа, которые делают ООП таким мощным инструментом.

### 4.1. Инкапсуляция (Encapsulation)

**Идея**: Объединение данных и методов в классе и **сокрытие** деталей реализации. Доступ к данным должен осуществляться через публичные методы, а не напрямую. Это защищает данные от некорректного изменения.

В Python для этого используются соглашения об именовании:
*   `_имя_атрибута` (одно подчеркивание): "Защищенный" атрибут. Это сигнал: "Не трогайте это напрямую снаружи, это для внутреннего использования".
*   `__имя_атрибута` (два подчеркивания): "Приватный" атрибут. Python изменяет его имя (`_ИмяКласса__имя_атрибута`), чтобы избежать случайного доступа извне.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        # Прячем баланс, чтобы его нельзя было сделать отрицательным напрямую
        self.__balance = balance 

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Счет пополнен на {amount}. Новый баланс: {self.__balance}")
        else:
            print("Сумма пополнения должна быть положительной.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Со счета списано {amount}. Новый баланс: {self.__balance}")
        else:
            print("Недостаточно средств или некорректная сумма списания.")

    def get_balance(self):
        # Предоставляем "безопасный" метод для получения информации о балансе
        print(f"Текущий баланс: {self.__balance}")
        
```

```python
account = BankAccount("Иван Петров", 1000)
account.get_balance()

# Попытка изменить баланс напрямую не сработает как ожидается
account.__balance = -5000
print("Попытались установить баланс -5000 напрямую...")
account.get_balance() # Баланс не изменился!

# Работаем через публичные методы
account.withdraw(500)
account.withdraw(1000) # Некорректная операция
```

### 4.2. Наследование (Inheritance)

**Идея**: Создание нового класса (дочернего) на основе существующего (родительского). Дочерний класс "наследует" все атрибуты и методы родительского, но может добавлять свои собственные или изменять (переопределять) унаследованные. Это позволяет избежать дублирования кода.
Отношение "является" (is-a). Например, `Собака` *является* `Животным`.

```python
# 1. Родительский класс
class Animal:
    def __init__(self, name):
        self.name = name
    def eat(self):
        print(f"{self.name} ест.")

# 2. Дочерний класс Dog наследует от Animal
class Dog(Animal):
    def bark(self): # Новый, собственный метод
        print(f"{self.name} говорит: Гав!")
        
# 3. Дочерний класс Cat наследует от Animal
class Cat(Animal):
    def meow(self): # Новый, собственный метод
        print(f"{self.name} говорит: Мяу!")

dog = Dog("Рекс")
cat = Cat("Барсик")

dog.eat() # Метод унаследован от Animal
dog.bark() # Собственный метод Dog

cat.eat() # Метод унаследован от Animal
cat.meow() # Собственный метод Cat
```

### 4.3. Полиморфизм (Polymorphism)

**Идея**: "Много форм". Это способность объектов разных классов, имеющих общий родительский класс, отвечать на один и тот же вызов метода, но выполнять разные действия.

**Переопределение методов (Method Overriding)**
Часто дочернему классу нужно изменить поведение унаследованного метода. Для этого метод просто определяется заново с тем же именем. Функция `super()` используется для вызова родительской версии метода.

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def make_sound(self):
        print("Издает какой-то животный звук...")

class Dog(Animal):
    # Переопределяем метод родителя
    def make_sound(self):
        print(f"{self.name} громко лает: ГАВ-ГАВ!")

class Cat(Animal):
    # Переопределяем метод родителя
    def make_sound(self):
        print(f"{self.name} нежно мурлычет: Мяууу...")

class Cow(Animal): # Эта корова не переопределяет метод
    pass

# Создаем список из разных животных
animals = [Dog("Рекс"), Cat("Мурка"), Cow("Зорька"), Animal("Нечто")]

# Полиморфизм в действии: мы вызываем один и тот же метод, но результат разный!
for animal in animals:
    animal.make_sound()
```

## 5. Продвинутые возможности

### 5.1. Документирование классов: Docstrings

**Docstring** (строка документации) — это строка в тройных кавычках `"""..."""`, которая идет сразу после определения класса или метода. Она описывает, что делает код. Это стандарт хорошего тона и основа для автоматической генерации документации. Функция `help()` показывает именно docstring.

```python
class Car:
    """Класс, представляющий легковой автомобиль.

    Attributes:
        brand (str): Марка автомобиля.
        model (str): Модель автомобиля.
        _current_speed (int): Текущая скорость (защищенный атрибут).
    """
    def __init__(self, brand, model):
        """Инициализирует автомобиль с маркой и моделью."""
        self.brand = brand
        self.model = model
        self._current_speed = 0

    def accelerate(self, value):
        """Увеличивает скорость автомобиля.

        Args:
            value (int): Значение, на которое увеличить скорость.
        """
        self._current_speed += value
        print(f"Скорость увеличилась до {self._current_speed} км/ч.")

# Посмотрим, как это работает с функцией help()
help(Car)
```

### 5.2. "Магические" методы (Dunder methods)

Методы, имена которых начинаются и заканчиваются двойным подчеркиванием (double underscore -> dunder), например, `__init__`, `__str__`, `__len__`, позволяют интегрировать наши объекты в язык Python. Они определяют, как объекты будут реагировать на встроенные операции (`print()`, `len()`, `+` и др.).

*   `__str__(self)`: Для `print(obj)` и `str(obj)`. Должен возвращать "красивую" строку для пользователя.
*   `__repr__(self)`: Для вывода объекта в консоли, отладке. Должен возвращать однозначное представление, в идеале — код для воссоздания объекта.
*   `__len__(self)`: Для `len(obj)`. 
*   `__eq__(self, other)`: Для оператора сравнения `==`.

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        
    # Как объект будет выглядеть при вызове print()
    def __str__(self):
        return f'Книга "{self.title}" от автора {self.author}'

    # Как объект будет выглядеть, если просто вывести его в консоли
    def __repr__(self):
        return f"Book(title='{self.title}', author='{self.author}', pages={self.pages})"

    # Как будет работать len()
    def __len__(self):
        return self.pages
        
    # Как будет работать ==
    def __eq__(self, other):
        if not isinstance(other, Book):
            return False
        return self.title == other.title and self.author == other.author

book1 = Book("Война и мир", "Л.Н. Толстой", 1300)
book2 = Book("Анна Каренина", "Л.Н. Толстой", 864)
book3 = Book("Война и мир", "Л.Н. Толстой", 1225) # Другое издание

print(book1) # __str__
print(f"В книге {len(book1)} страниц.") # __len__

print(f"book1 и book2 одинаковы? {book1 == book2}") # __eq__ -> False
print(f"book1 и book3 одинаковы? {book1 == book3}") # __eq__ -> True

print([book1, book2]) # в контейнерах используется __repr__
```
