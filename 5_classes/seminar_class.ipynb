### Задание 3.7: Наследование и переопределение с `super()`

1. Создайте класс `Employee` с методом `get_info()`, возвращающим `"Работник: [Имя]"`.
2. Создайте дочерний класс `Manager`. Добавьте ему список подчиненных `_subordinates`.
3. Переопределите `get_info()` в `Manager`, чтобы он вызывал родительскую версию через `super()` и добавлял к результату `", Подчиненных: [Количество]"`.

```python
# ----- ВАШ КОД ЗДЕСЬ ----- (Задача 3.7)

# -------------------------
```

```python
# ----- Тесты для Задачи 3.7 -----
try:
    e1 = Employee("Петр", 50000)
    m = Manager("Светлана", 100000)
    assert m.get_info() == "Работник: Светлана, Подчиненных: 0"
    m.add_subordinate(e1)
    assert m.get_info() == "Работник: Светлана, Подчиненных: 1"
    print("✅ Все тесты для Задачи 3.7 пройдены!")
except NameError:
    print("❌ Тест для Задачи 3.7 провален: Классы 'Employee' или 'Manager' не определены.")
except Exception as e:
    print(f"❌ Тест для Задачи 3.7 провален: {e}")
```
