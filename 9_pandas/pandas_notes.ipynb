# Большой и подробный конспект по Pandas

**Pandas** — это высокоуровневая библиотека Python для анализа данных. Она предоставляет быстрые, гибкие и выразительные структуры данных, разработанные для того, чтобы сделать работу с "реляционными" или "помеченными" данными простой и интуитивно понятной.

Основные структуры данных в Pandas:
1.  **Series** — одномерный маркированный массив, способный хранить данные любого типа.
2.  **DataFrame** — двумерная маркированная структура данных с колонками, которые могут иметь разные типы. Её можно представить как таблицу, похожую на лист Excel или SQL-таблицу.

Давайте начнём с основ и будем двигаться к более сложным темам.

## 1. Основы: Series и DataFrame

Прежде чем начать, импортируем библиотеку. Общепринятое сокращение — `pd`.

```python
import pandas as pd
import numpy as np
```

### 1.1. Series

**Series** — это одномерный массив, похожий на список или столбец в таблице. У каждого элемента есть метка, которая называется **индекс**.

#### Создание Series

```python
# Из списка Python
data_list = [10, 20, 30, 40, 50]
s1 = pd.Series(data_list)
print("Series из списка:")
print(s1)

# По умолчанию индекс - это числа от 0. Можно задать свой индекс:
s2 = pd.Series(data_list, index=['a', 'b', 'c', 'd', 'e'])
print("\nSeries с заданным индексом:")
print(s2)

# Из словаря Python
# Ключи словаря автоматически становятся индексами.
data_dict = {'Москва': 12.5, 'Санкт-Петербург': 5.4, 'Новосибирск': 1.6}
s3 = pd.Series(data_dict)
print("\nSeries из словаря (население в млн):")
print(s3)
```

#### Атрибуты и доступ к данным в Series

У Series есть значения (`values`) и индекс (`index`).

```python
print("Значения Series s2:", s2.values)
print("Индекс Series s2:", s2.index)

# Доступ к элементам можно получать как по числовому индексу (как в списке), так и по метке индекса (как в словаре).

# По метке индекса
print("\nЭлемент с индексом 'b':", s2['b'])

# По числовому индексу (позиции)
print("Элемент на позиции 0:", s2[0])

# Срез по меткам (включая конечный элемент!)
print("\nСрез от 'b' до 'd':")
print(s2['b':'d'])

# Срез по числовым индексам (НЕ включая конечный элемент, как в списках)
print("\nСрез от 1 до 4 (не включая 4):")
print(s2[1:4])
```

### 1.2. DataFrame

**DataFrame** — это двумерная структура, похожая на таблицу. У неё есть и индекс строк, и индекс столбцов. Каждый столбец DataFrame — это объект Series.

#### Создание DataFrame

```python
# Самый частый способ: из словаря, где ключи - названия столбцов, а значения - списки.
data = {
    'Имя': ['Анна', 'Борис', 'Виктор', 'Галина', 'Дмитрий'],
    'Возраст': [28, 34, 29, 42, 31],
    'Город': ['Москва', 'СПб', 'Екатеринбург', 'Москва', 'Новосибирск'],
    'Зарплата': [120000, 95000, 110000, 150000, 80000]
}

df = pd.DataFrame(data)

print("Созданный DataFrame:")
df # В Jupyter Notebook можно просто написать имя переменной для красивого вывода
```

Как и в Series, мы можем задать свой индекс.

```python
df_with_index = pd.DataFrame(data, index=['emp1', 'emp2', 'emp3', 'emp4', 'emp5'])
print("DataFrame с кастомным индексом:")
df_with_index
```

Редко когда данные создаются вручную. Обычно их загружают из файлов. Самый популярный формат — CSV (Comma-Separated Values).

## 4. Извлечение информации из датафреймов (индексация и срезы)

Это одна из самых важных и мощных частей Pandas. Существует несколько способов выбрать нужные вам данные.
Для примеров используем наш первый DataFrame `df_people`.

```python
data = {
    'Имя': ['Анна', 'Борис', 'Виктор', 'Галина', 'Дмитрий'],
    'Возраст': [28, 34, 29, 42, 31],
    'Город': ['Москва', 'СПб', 'Екатеринбург', 'Москва', 'Новосибирск'],
    'Зарплата': [120000, 95000, 110000, 150000, 80000]
}
df_people = pd.DataFrame(data, index=['e1', 'e2', 'e3', 'e4', 'e5'])
df_people
```

### 4.1. Выбор столбцов

```python
# Выбор одного столбца (результат - Series)
ages = df_people['Возраст']
print("Тип данных при выборе одного столбца:", type(ages))
print(ages)

# Альтернативный синтаксис (работает, если имя столбца не содержит пробелов и не совпадает с методами df)
# print(df_people.Возраст)

# Выбор нескольких столбцов (результат - DataFrame)
# Обратите внимание на двойные квадратные скобки: df[['col1', 'col2']]
name_and_salary = df_people[['Имя', 'Зарплата']]
print("\nТип данных при выборе нескольких столбцов:", type(name_and_salary))
display(name_and_salary)
```

### 4.2. Выбор строк: `.loc` и `.iloc`

Это предпочтительный способ выбора строк. Основное различие:
-   `.loc` (location) - работает с **метками** индекса и названиями столбцов.
-   `.iloc` (integer location) - работает с **числовыми позициями** (начиная с 0).

**Формат: `df.loc[строки, столбцы]`**

#### Использование `.loc`

```python
# Выбор одной строки по метке индекса (результат - Series)
print("Строка с индексом 'e3':")
print(df_people.loc['e3'])

# Выбор нескольких строк по списку меток (результат - DataFrame)
print("\nСтроки с индексами 'e1' и 'e4':")
display(df_people.loc[['e1', 'e4']])

# Срез по меткам (включая последнюю!)
print("\nСрез строк от 'e2' до 'e4':")
display(df_people.loc['e2':'e4'])

# Выбор конкретных строк и столбцов одновременно
print("\nИмя и город для сотрудников 'e1' и 'e5':")
display(df_people.loc[['e1', 'e5'], ['Имя', 'Город']])

# Выбор всех строк, но определенных столбцов
print("\nВсе сотрудники, но только столбцы 'Возраст' и 'Зарплата':")
display(df_people.loc[:, ['Возраст', 'Зарплата']])
```

#### Использование `.iloc`

```python
# Выбор одной строки по её позиции (индекс 0 - первая строка)
print("Первая строка (позиция 0):")
print(df_people.iloc[0])

# Выбор нескольких строк по списку позиций
print("\nСтроки на позициях 0, 2, 4:")
display(df_people.iloc[[0, 2, 4]])

# Срез по позициям (НЕ включая последнюю, как в списках Python)
print("\nСрез строк с 1 по 4 (не включая 4-ю):")
display(df_people.iloc[1:4])

# Выбор конкретных строк и столбцов по их позициям
# Строки 0 и 3, столбцы 0 (Имя) и 2 (Город)
print("\nСтроки 0,3 и столбцы 0,2:")
display(df_people.iloc[[0, 3], [0, 2]])

# Выбор определенной ячейки
cell_value = df_people.iloc[1, 3] # Вторая строка, четвертый столбец
print("\nЗначение в ячейке [1, 3]:", cell_value)
```

## 5. Фильтрация данных (условный выбор)

Это особый, очень мощный вид извлечения данных, основанный на условиях. Он позволяет отбирать строки, которые удовлетворяют определенным критериям.

#### Принцип работы:
1.  Создается "булева маска" — это Series из `True` и `False` той же длины, что и DataFrame.
2.  Эта маска передается в `df[...]` или `df.loc[...]`.
3.  DataFrame возвращает только те строки, где в маске стоит `True`.

```python
# Шаг 1: Создаем булеву маску
mask_age_gt_30 = df_people['Возраст'] > 30
print("Булева маска (Возраст > 30):")
print(mask_age_gt_30)

# Шаг 2 и 3: Применяем маску
print("\nСотрудники старше 30 лет:")
display(df_people[mask_age_gt_30])
```

### 5.1. Комбинирование условий

Для комбинирования условий используются побитовые операторы:
-   `&` для логического **И** (AND)
-   `|` для логического **ИЛИ** (OR)
-   `~` для логического **НЕ** (NOT)

**ВАЖНО:** Каждое условие нужно заключать в круглые скобки `()` из-за приоритета операторов в Python.

```python
# Фильтр: сотрудники из Москвы И старше 30 лет
print("Сотрудники из Москвы старше 30:")
display(df_people[(df_people['Город'] == 'Москва') & (df_people['Возраст'] > 30)])

# Фильтр: сотрудники из СПб ИЛИ с зарплатой > 100000
print("\nСотрудники из СПб или с зарплатой > 100 000:")
display(df_people[(df_people['Город'] == 'СПб') | (df_people['Зарплата'] > 100000)])
```

### 5.2. Дополнительные методы для фильтрации

- `.isin()`: проверяет, входит ли значение в заданный список.
- `.isnull()` / `.notnull()`: проверяет на наличие пропусков (NaN).

```python
# Фильтр: сотрудники из Москвы или Новосибирска
cities_of_interest = ['Москва', 'Новосибирск']
print("Сотрудники из Москвы или Новосибирска:")
display(df_people[df_people['Город'].isin(cities_of_interest)])

# Используем DataFrame с продуктами, где есть пропуски
print("\nТовары с неизвестным количеством на складе:")
display(df[df['stock'].isnull()])

# Товары, у которых количество на складе указано
print("\nТовары с известным количеством на складе:")
display(df[df['stock'].notnull()])
```

## 6. Добавление и удаление данных

### 6.1. Добавление столбцов

Это одна из самых частых операций. Новый столбец можно создать на основе существующих или просто присвоить ему скалярное значение.

```python
df_people_mod = df_people.copy() # Работаем с копией

# Добавление столбца с постоянным значением
df_people_mod['Страна'] = 'Россия'
display(df_people_mod)

# Добавление вычисляемого столбца
# Например, зарплата после уплаты налога 13% (НДФЛ)
df_people_mod['ЗП на руки'] = df_people_mod['Зарплата'] * 0.87
display(df_people_mod)

# Можно добавить столбец с помощью pd.Series, главное, чтобы индекс совпадал
bonuses = pd.Series([10000, 5000, 7000, 15000, 4000], index=['e1', 'e2', 'e3', 'e4', 'e5'])
df_people_mod['Бонус'] = bonuses
display(df_people_mod)
```

Для удаления используется метод `.drop()`. Он может удалять как строки, так и столбцы.

```python
# Удаление столбца
# axis=1 - ОБЯЗАТЕЛЬНО для удаления столбцов
# Метод .drop() по умолчанию возвращает новый DataFrame, не изменяя исходный.
df_dropped_col = df_people_mod.drop('Бонус', axis=1)
print("После удаления столбца 'Бонус':")
display(df_dropped_col)

# Удаление нескольких столбцов
df_dropped_cols = df_people_mod.drop(['Страна', 'Бонус'], axis=1)
print("\nПосле удаления столбцов 'Страна' и 'Бонус':")
display(df_dropped_cols)

# Удаление строк по индексу
# axis=0 - для строк (это значение по умолчанию)
df_dropped_rows = df_people_mod.drop(['e2', 'e4'])
print("\nПосле удаления строк 'e2' и 'e4':")
display(df_dropped_rows)

# Чтобы изменить DataFrame на месте, используется параметр inplace=True
df_people_inplace = df_people_mod.copy()
df_people_inplace.drop('Страна', axis=1, inplace=True)
print("\nИзменение DataFrame на месте с inplace=True:")
display(df_people_inplace)
```

## 7. Комбинирование датафреймов

Pandas предоставляет несколько способов объединения данных из разных DataFrame.

### 7.1. Конкатенация с `pd.concat()`

`pd.concat()` "склеивает" датафреймы вместе. Может склеивать как по вертикали (добавляя строки), так и по горизонтали (добавляя столбцы).

```python
# Создадим два небольших DataFrame для примера
df1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']})
df2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']})

# Вертикальная конкатенация (axis=0, по умолчанию)
v_concat = pd.concat([df1, df2])
print("Вертикальная конкатенация:")
display(v_concat)
# Обратите внимание, что индексы дублируются. Это можно исправить.
v_concat_reindex = pd.concat([df1, df2], ignore_index=True)
print("\nВертикальная конкатенация с ignore_index=True:")
display(v_concat_reindex)

# Горизонтальная конкатенация (axis=1)
df3 = pd.DataFrame({'C': ['C0', 'C1'], 'D': ['D0', 'D1']})
h_concat = pd.concat([df1, df3], axis=1)
print("\nГоризонтальная конкатенация:")
display(h_concat)
```

### 7.2. Слияние с `pd.merge()`

`pd.merge()` — это аналог SQL `JOIN`. Он объединяет DataFrame на основе значений в одном или нескольких общих столбцах (ключах).

Основные типы слияния (параметр `how`):
-   `inner` (по умолчанию): оставляет только те строки, где ключи есть в **обоих** DataFrame.
-   `outer`: оставляет **все** строки из обоих DataFrame. Если в одном из них нет ключа, значения в соответствующих столбцах другого заполняются `NaN`.
-   `left`: оставляет **все** строки из **левого** DataFrame и соответствующие им из правого.
-   `right`: оставляет **все** строки из **правого** DataFrame и соответствующие им из левого.

```python
# Создадим два DataFrame для примера
left = pd.DataFrame({
   'key': ['K0', 'K1', 'K2', 'K3'],
   'A': ['A0', 'A1', 'A2', 'A3'],
   'B': ['B0', 'B1', 'B2', 'B3']
})

right = pd.DataFrame({
   'key': ['K0', 'K1', 'K4', 'K5'],
   'C': ['C0', 'C1', 'C4', 'C5'],
   'D': ['D0', 'D1', 'D4', 'D5']
})
display("Левый DF:", left)
display("Правый DF:", right)

# Inner Join
inner_join = pd.merge(left, right, on='key', how='inner')
display("Inner Join:", inner_join)

# Outer Join
outer_join = pd.merge(left, right, on='key', how='outer')
display("Outer Join (NaN там, где нет совпадений):", outer_join)

# Left Join
left_join = pd.merge(left, right, on='key', how='left')
display("Left Join (все ключи из левого DF):", left_join)

# Right Join
right_join = pd.merge(left, right, on='key', how='right')
display("Right Join (все ключи из правого DF):", right_join)
```

### 8.1. Работа с пропущенными данными

```python
# Найти пропуски: isnull() или isna()
print("Маска пропусков:")
display(df.isnull())

# Посчитать количество пропусков в каждом столбце
print("\nКоличество пропусков по столбцам:")
print(df.isnull().sum())

# Удаление строк с пропусками: dropna()
df_no_na = df.dropna()
print("\nDataFrame после удаления строк с пропусками:")
display(df_no_na)

# Заполнение пропусков: fillna()
# Можно заполнить нулем
df_filled_zero = df.fillna(0)
print("\nDataFrame после заполнения пропусков нулем:")
display(df_filled_zero)

# Более продвинутый способ: заполнить средним значением по столбцу
mean_stock = df['stock'].mean()
df_filled_mean = df.copy()
df_filled_mean['stock'].fillna(mean_stock, inplace=True)
print(f"\nDataFrame после заполнения пропусков средним значением ({mean_stock:.1f}):")
display(df_filled_mean)
```

### 8.2. Группировка и агрегация: `groupby()`

`groupby()` — один из самых мощных инструментов в Pandas. Он позволяет реализовать парадигму **Split-Apply-Combine** (Разделить-Применить-Объединить):

1.  **Split:** Данные делятся на группы на основе некоторого критерия (например, по категориям товаров).
2.  **Apply:** К каждой группе независимо применяется некоторая функция (например, `sum`, `mean`, `count`).
3.  **Combine:** Результаты применения функции объединяются в новую структуру данных.


```python
# Давайте сгруппируем товары по категориям и посчитаем среднюю цену в каждой
avg_price_by_category = df.groupby('category')['price'].mean()
print("Средняя цена по категориям:")
print(avg_price_by_category)

# Можно применять несколько агрегаций одновременно с помощью .agg()
agg_by_category = df.groupby('category').agg(
    total_stock=('stock', 'sum'),
    avg_price=('price', 'mean'),
    product_count=('product_id', 'count')
)
print("\nАгрегированные данные по категориям:")
display(agg_by_category)
```

### 8.3. Сортировка данных

-   `.sort_values()` - сортировка по значениям в одном или нескольких столбцах.
-   `.sort_index()` - сортировка по индексу.

```python
# Сортировка по одному столбцу (по возрастанию цены)
df_sorted_price = df.sort_values(by='price')
print("Сортировка по цене (возрастание):")
display(df_sorted_price)

# Сортировка по убыванию
df_sorted_price_desc = df.sort_values(by='price', ascending=False)
print("\nСортировка по цене (убывание):")
display(df_sorted_price_desc)

# Сортировка по нескольким столбцам
# Сначала по категории (алфавитный порядок), затем внутри каждой категории по цене (убывание)
df_sorted_multi = df.sort_values(by=['category', 'price'], ascending=[True, False])
print("\nСортировка по категории (возр.) и цене (убыв.):")
display(df_sorted_multi)
```

### 8.4. Применение функций: `apply()` и `map()`

Иногда нужно применить к данным более сложную логику, чем стандартные агрегации.

-   `.apply()`: может применяться к столбцам/строкам DataFrame или к элементам Series.
-   `.map()`: применяется поэлементно только к Series.

```python
# Используем .map() на Series для замены значений
category_map = {
    'Молочные продукты': 'Dairy',
    'Хлебобулочные изделия': 'Bakery',
    'Фрукты': 'Fruits',
    'Овощи': 'Vegetables',
    'Мясо': 'Meat'
}
df['category_en'] = df['category'].map(category_map)
print("DataFrame с новым столбцом 'category_en' через .map():")
display(df)

# Используем .apply() на столбце (Series) с lambda-функцией
# Присвоим статус 'дорогой' или 'дешевый' в зависимости от цены
df['price_status'] = df['price'].apply(lambda x: 'Дорогой' if x > 100 else 'Дешевый')
print("\nDataFrame со статусом цены, созданным через .apply():")
display(df[['product_name', 'price', 'price_status']])

# .apply() по строкам (axis=1)
# Это менее эффективно, чем векторизованные операции, но иногда необходимо
# Создадим описание товара на основе нескольких столбцов
def create_description(row):
    return f"{row['product_name']} из категории '{row['category_en']}' стоит {row['price']} руб."

df['description'] = df.apply(create_description, axis=1)
print("\nDataFrame с полным описанием, созданным через .apply(axis=1):")
display(df[['description']])
```
