- **PassengerId**: Уникальный ID пассажира
- **Survived**: Выжил ли пассажир (0 = Нет, 1 = Да)
- **Pclass**: Класс билета (1 = 1-й, 2 = 2-й, 3 = 3-й)
- **Name**: Имя
- **Sex**: Пол
- **Age**: Возраст в годах

- **Parch**: Количество родителей, детей на борту
- **Ticket**: Номер билета
- **Fare**: Стоимость проезда (тариф)
- **Cabin**: Номер каюты
- **Embarked**: Порт посадки (C = Шербур, Q = Квинстаун, S = Саутгемптон)
