## Задача 1 (Лёгкая): Первый запрос

**Задание:**
Используя библиотеку `requests`, сделайте GET-запрос на URL `https://httpbin.org/get`. Сохраните результат (объект `Response`) в переменную `response`.

````python
import requests

# ▼▼▼ ВАШ КОД ЗДЕСЬ ▼▼▼
url = 'https://httpbin.org/get'
response = None # Замените None на ваш код

# ▲▲▲ ВАШ КОД ЗДЕСЬ ▲▲▲

print(f"Статус-код: {response.status_code}")
print(f"Тип контента: {response.headers.get('Content-Type')}")
````
