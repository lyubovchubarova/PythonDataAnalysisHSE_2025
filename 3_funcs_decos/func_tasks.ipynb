#### Решение и тесты

```python
calls = 0
@retry(max_attempts=3)
def succeed_on_third_try():
    global calls
    calls += 1
    if calls < 3:
        raise ValueError("Неудача")
    return "Успех!"

assert succeed_on_third_try() == "Успех!"
assert calls == 3

calls_fail = 0
@retry(max_attempts=4)
def always_fail():
    global calls_fail
    calls_fail += 1
    raise RuntimeError(f"Ошибка номер {calls_fail}")

try:
    always_fail()
except RuntimeError as e:
    assert calls_fail == 4
    assert str(e) == "Ошибка номер 4"
else:
    assert False, "RuntimeError не была вызвана"

print("Задача 5.3: тесты пройдены! ✅")
```
