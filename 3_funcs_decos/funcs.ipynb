# Функции и лямбда-функции в Python: Конспект-шпаргалка

## Часть 1: Обычные функции (`def`)

### 1.1. Что такое функция и зачем она нужна?

**Функция** — это именованный блок кода, который выполняет определённую задачу. Вы можете вызывать этот блок кода по имени в любом месте программы, сколько угодно раз.

**Зачем использовать функции?**
1.  **Повторное использование кода (Don't Repeat Yourself - DRY)**: Вместо того чтобы копировать и вставлять один и тот же код, вы пишете его один раз в функции и вызываете по необходимости.
2.  **Организация кода**: Функции разбивают большую, сложную программу на маленькие, управляемые и понятные части.
3.  **Читаемость**: Хорошо названная функция (например, `calculate_average_grade`) сразу говорит о том, что она делает.

Представьте функцию как кулинарный рецепт: у неё есть название (`"Борщ"`), список ингредиентов (аргументы) и последовательность шагов (тело функции).

### 1.2. Как создать функцию?

Для создания функции используется ключевое слово `def`.

**Синтаксис:**
```python
def имя_функции(аргументы):
    # Тело функции: код, который что-то делает
    # ...
    return результат # Необязательно
```

**Пример простой функции:**

````python
def say_hello():
    print("Привет, мир!")

# Вызов функции
say_hello()
````

### 1.3. Аргументы функции (ингредиенты)

Аргументы — это данные, которые мы передаём в функцию для обработки.

#### а) Позиционные аргументы
Передаются в функцию в определенном порядке. Порядок важен!

````python
def introduce(name, city):
    print(f"Меня зовут {name}, я из города {city}.")

# Правильный вызов
introduce("Алекс", "Москва") 

# Неправильный вызов (аргументы перепутаны)
introduce("Москва", "Алекс")
````

#### б) Именованные (ключевые) аргументы
Передаются в функцию с указанием имени аргумента. Порядок не важен.

````python
def introduce(name, city):
    print(f"Меня зовут {name}, я из города {city}.")

# Порядок не важен, результат будет верным
introduce(city="Санкт-Петербург", name="Мария")
````

#### в) Аргументы со значениями по умолчанию
Можно задать значение для аргумента на случай, если его не передали при вызове. Такие аргументы должны идти после всех позиционных аргументов.

````python
def greet(name, greeting="Привет"):
    print(f"{greeting}, {name}!")

# Вызываем с аргументом по умолчанию
greet("Иван") 

# Переопределяем аргумент по умолчанию
greet("Анна", greeting="Добрый день")
````

### 1.4. Возвращение значений (`return`)

Часто функция должна не просто напечатать что-то, а **вернуть** результат своей работы, чтобы его можно было использовать дальше. Для этого используется оператор `return`.

````python
def calculate_area(width, height):
    """Эта функция вычисляет площадь прямоугольника."""
    area = width * height
    return area

# Вызываем функцию и сохраняем результат в переменную
rectangle_area = calculate_area(10, 5) 

print(f"Площадь прямоугольника: {rectangle_area}")
print(f"Двойная площадь: {rectangle_area * 2}")
````

Если в функции нет `return`, она неявно возвращает `None`.

````python
def do_nothing():
    x = 1 + 1

result = do_nothing()
print(result)
````

### 1.5. Строки документации (Docstrings)

Это специальные комментарии в тройных кавычках `"""..."""` сразу после определения функции. Они описывают, что делает функция. Это хорошая практика, которая помогает другим (и вам в будущем) понять ваш код.

````python
def calculate_area(width, height):
    """
    Вычисляет площадь прямоугольника.

    Аргументы:
        width (int, float): Ширина прямоугольника.
        height (int, float): Высота прямоугольника.

    Возвращает:
        int, float: Площадь прямоугольника.
    """
    return width * height

# Посмотреть документацию можно так:
help(calculate_area)
````

## Часть 2: Лямбда-функции

### 2.1. Что такое лямбда-функция?
**Лямбда-функция** — это маленькая **анонимная** (безымянная) функция, которая может содержать только **одно выражение**.

Её используют, когда нужна простая функция на один раз, и нет смысла объявлять её через `def`.

### 2.2. Синтаксис

```python
lambda аргументы: выражение
```
- `lambda` — ключевое слово.
- `аргументы` — то, что функция принимает.
- `выражение` — то, что функция возвращает.

**Сравнение `def` и `lambda`:**

````python
# Обычная функция
def add(x, y):
    return x + y

# Эквивалентная ей лямбда-функция
adder_lambda = lambda x, y: x + y

print(f"Результат def: {add(5, 3)}")
````

### 2.3. Где и зачем это использовать?

Основная сила лямбда-функций — в их использовании "на лету" в качестве аргументов для других, более высокоуровневых функций, таких как `map`, `filter` и `sorted`.

#### а) Использование с `map()`
`map(функция, итерируемый_объект)` применяет `функцию` к каждому элементу `итерируемого_объекта`.

**Задача**: Возвести в квадрат все числа в списке.

````python
numbers = [1, 2, 3, 4, 5]

# Вместо того чтобы писать цикл или отдельную функцию...
squared_numbers = list(map(lambda x: x * x, numbers))

print(squared_numbers)
````

#### б) Использование с `filter()`
`filter(функция, итерируемый_объект)` оставляет только те элементы, для которых `функция` вернула `True`.

**Задача**: Оставить в списке только четные числа.

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
````

#### в) Использование с `sorted()`
`sorted()` может принимать необязательный аргумент `key`, который указывает на функцию, используемую для определения порядка сортировки.

**Задача**: Отсортировать список словарей со студентами по их возрасту.

````python
from pprint import pprint

students = [
    {'name': 'Иван', 'age': 22},
    {'name': 'Мария', 'age': 19},
    {'name': 'Петр', 'age': 20}
]

# Ключ сортировки - возраст студента
sorted_students = sorted(students, key=lambda student: student['age'])

pprint(sorted_students)
````

## Заключение: `def` vs `lambda`

| Характеристика          | `def` (обычная функция)                                | `lambda` (лямбда-функция)                              |
| ----------------------- | ------------------------------------------------------ | ------------------------------------------------------ |
| **Наличие имени**       | Всегда имеет имя                                       | Анонимная (безымянная)                                 |
| **Сложность**           | Может содержать много строк кода, циклы, условия и т.д. | Только одно выражение                                  |
| **Возврат значения**    | Использует `return` для возврата, может не возвращать ничего | Всегда неявно возвращает результат своего выражения    |
| **Основное применение** | Создание многократно используемых, сложных блоков кода   | Быстрые одноразовые функции для `map`, `filter`, `sorted` |
